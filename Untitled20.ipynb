# Future Sales Prediction Workflow
In this notebook, we'll build a machine learning model to predict future product sales using a regression approach. We will generate simulated sales data with trend and seasonal dynamics, engineer lag/time features, train a Random Forest model, evaluate its performance, and forecast sales into the future.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import datetime
```

## 1. Create a Synthetic Sales Dataset
We generate daily sales data spanning 3 years, embedding realistic trends, weekly cycles (higher sales on weekends), and annual seasonality.

```python
# Generate synthetic sales data
np.random.seed(42)
start_date = datetime.date(2021, 1, 1)
end_date = datetime.date(2023, 12, 31)
dates = pd.date_range(start=start_date, end=end_date, freq='D')

# Dynamic components
trend = np.linspace(50, 150, len(dates))  # General upward trend
weekly_seasonality = np.array([1.2, 0.9, 0.85, 0.9, 1.0, 1.3, 1.4])[dates.dayofweek]  # Higher on weekends
yearly_seasonality = 1 + 0.3 * np.sin(2 * np.pi * dates.dayofyear / 365.25)  # Annual peak
noise = np.random.normal(0, 15, len(dates))

# Calculate final sales
sales = (trend * weekly_seasonality * yearly_seasonality) + noise
sales = np.clip(sales, a_min=10, a_max=None) # Keep values positive

df_sales = pd.DataFrame({'Date': dates, 'Sales': sales})
print(f"Dataset successfully created. Shape: {df_sales.shape}")
display(df_sales.head())
```

## 2. Feature Engineering
To make this a machine learning problem, we extract date-based features (month, day of week, day of year) and historical lag features (previous day's sales, sales from 7 days ago).

```python
# Extract temporal features
df_sales['Month'] = df_sales['Date'].dt.month
df_sales['DayOfWeek'] = df_sales['Date'].dt.dayofweek
df_sales['DayOfYear'] = df_sales['Date'].dt.dayofyear
df_sales['Year'] = df_sales['Date'].dt.year

# Create historical lag features
df_sales['Lag_1'] = df_sales['Sales'].shift(1)
df_sales['Lag_7'] = df_sales['Sales'].shift(7)
df_sales['Rolling_Mean_7'] = df_sales['Sales'].shift(1).rolling(window=7).mean()

# Drop rows with NaN values generated due to shifting
df_sales = df_sales.dropna().reset_index(drop=True)

display(df_sales.head())
```

## 3. Model Training & Evaluation
We split the data into training and testing sets (using a chronological split to mimic realistic time-series validation), train a Random Forest regressor, and evaluate the performance.

```python
# Split features and target
features = ['Month', 'DayOfWeek', 'DayOfYear', 'Year', 'Lag_1', 'Lag_7', 'Rolling_Mean_7']
X = df_sales[features]
y = df_sales['Sales']

# Chronological train-test split (80% Train, 20% Test)
split_idx = int(len(df_sales) * 0.8)
X_train, X_test = X.iloc[:split_idx], X.iloc[split_idx:]
y_train, y_test = y.iloc[:split_idx], y.iloc[split_idx:]
test_dates = df_sales['Date'].iloc[split_idx:]

# Train Random Forest Regressor
model = RandomForestRegressor(n_estimators=100, random_state=42)
model.fit(X_train, y_train)

# Predict and evaluate
y_pred = model.predict(X_test)
mae = mean_absolute_error(y_test, y_pred)
mse = mean_squared_error(y_test, y_pred)
rmse = np.sqrt(mse)
r2 = r2_score(y_test, y_pred)

print("Model Evaluation Metrics:")
print(f"- Mean Absolute Error (MAE): {mae:.2f}")
print(f"- Root Mean Squared Error (RMSE): {rmse:.2f}")
print(f"- R-squared (R2 Score): {r2:.4f}")
```

## 4. Visualization of Results
Let's plot the actual vs predicted sales values to assess how well our model captures seasonal swings and the general trend.

```python
plt.figure(figsize=(14, 6))
plt.plot(test_dates, y_test.values, label='Actual Sales', color='tab:blue', alpha=0.8)
plt.plot(test_dates, y_pred, label='Predicted Sales', color='tab:orange', alpha=0.9, linestyle='--')
plt.title('Future Sales Prediction: Actual vs Predicted (Test Set)')
plt.xlabel('Date')
plt.ylabel('Sales Volume')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```
