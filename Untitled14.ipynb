### 1. Iris Flower Classification using Naive Bayes

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn import datasets
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score

# Load the Iris dataset
iris = datasets.load_iris()
X_iris = iris.data
y_iris = iris.target
target_names = iris.target_names

# Create a DataFrame for visualization
df_iris_class = pd.DataFrame(X_iris, columns=iris.feature_names)
df_iris_class['species'] = [target_names[i] for i in y_iris]
display(df_iris_class.head())
```

### 2. Split Data, Train Naive Bayes, and Evaluate

```python
# Split data into training and testing sets
X_train_ir, X_test_ir, y_train_ir, y_test_ir = train_test_split(X_iris, y_iris, test_size=0.2, random_state=42, stratify=y_iris)

# Initialize and train Gaussian Naive Bayes classifier
nb_classifier = GaussianNB()
nb_classifier.fit(X_train_ir, y_train_ir)

# Predict on the test set
y_pred_ir = nb_classifier.predict(X_test_ir)

# Calculate accuracy
acc_ir = accuracy_score(y_test_ir, y_pred_ir)
print(f"Classification Accuracy: {acc_ir * 100:.2f}%")
print("\nClassification Report:\n", classification_report(y_test_ir, y_pred_ir, target_names=target_names))
```

### 3. Confusion Matrix Visualization

```python
# Generate and visualize confusion matrix
cm_ir = confusion_matrix(y_test_ir, y_pred_ir)

plt.figure(figsize=(8, 6))
sns.heatmap(cm_ir, annot=True, fmt='d', cmap='Blues', xticklabels=target_names, yticklabels=target_names)
plt.title('Confusion Matrix - Iris Classification (Naive Bayes)')
plt.xlabel('Predicted Label')
plt.ylabel('True Label')
plt.show()
```
