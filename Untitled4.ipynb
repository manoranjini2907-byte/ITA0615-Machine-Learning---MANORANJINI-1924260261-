### Artificial Neural Network with Backpropagation

**Backpropagation** is the central algorithm for training neural networks. It works by calculating the gradient of the loss function with respect to the network weights, propagating the error backward from the output layer to the input layer using the chain rule.

We will implement a 3-layer neural network (Input, Hidden, Output) from scratch using `numpy` and train it on the classic non-linear **XOR** problem.

```python
import numpy as np

class NeuralNetwork:
    def __init__(self, input_size, hidden_size, output_size, learning_rate=0.1):
        self.input_size = input_size
        self.hidden_size = hidden_size
        self.output_size = output_size
        self.lr = learning_rate

        # Initialize weights and biases randomly
        self.W1 = np.random.uniform(size=(self.input_size, self.hidden_size))
        self.b1 = np.zeros((1, self.hidden_size))
        self.W2 = np.random.uniform(size=(self.hidden_size, self.output_size))
        self.b2 = np.zeros((1, self.output_size))

    def sigmoid(self, x):
        return 1 / (1 + np.exp(-x))

    def sigmoid_derivative(self, x):
        return x * (1 - x)

    def forward(self, X):
        # Forward propagation
        self.z1 = np.dot(X, self.W1) + self.b1
        self.a1 = self.sigmoid(self.z1)
        self.z2 = np.dot(self.a1, self.W2) + self.b2
        self.a2 = self.sigmoid(self.z2)
        return self.a2

    def backward(self, X, y, output):
        # Backpropagation
        # Error at output layer
        output_error = y - output
        output_delta = output_error * self.sigmoid_derivative(output)

        # Error at hidden layer
        hidden_error = output_delta.dot(self.W2.T)
        hidden_delta = hidden_error * self.sigmoid_derivative(self.a1)

        # Update weights and biases
        self.W2 += self.a1.T.dot(output_delta) * self.lr
        self.b2 += np.sum(output_delta, axis=0, keepdims=True) * self.lr
        self.W1 += X.T.dot(hidden_delta) * self.lr
        self.b1 += np.sum(hidden_delta, axis=0, keepdims=True) * self.lr

    def train(self, X, y, epochs=10000):
        for epoch in range(epochs):
            output = self.forward(X)
            self.backward(X, y, output)
            if epoch % 2000 == 0:
                loss = np.mean(np.square(y - output))
                print(f"Epoch {epoch} - Mean Squared Error Loss: {loss:.5f}")
```

```python
# Define XOR dataset
X = np.array([[0, 0], [0, 1], [1, 0], [1, 1]])
y = np.array([[0], [1], [1], [0]])

# Initialize the neural network
# 2 input features, 4 hidden units, 1 output layer node
ann = NeuralNetwork(input_size=2, hidden_size=4, output_size=1, learning_rate=0.2)

print("Starting Neural Network Training:")
ann.train(X, y, epochs=10000)

# Test the network
predictions = ann.forward(X)
print("\nFinal Test Results on XOR gate:")
for inputs, true_val, pred_val in zip(X, y, predictions):
    print(f"Input: {inputs} -> Target: {true_val[0]} -> Predicted: {pred_val[0]:.4f} (Class: {int(pred_val[0] > 0.5)})")
```
