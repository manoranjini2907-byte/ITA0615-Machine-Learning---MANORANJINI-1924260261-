### 1. Create a Synthetic House Price Dataset
We will construct a dataset with realistic features that influence house prices: **Square Footage (Size)**, **Number of Bedrooms**, **House Age**, and **Neighborhood Quality Index** (scaled from 1 to 10).

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.ensemble import GradientBoostingRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

# Set random seed for reproducibility
np.random.seed(42)
n_houses = 1000

# Generate synthetic features
size_sqft = np.random.normal(2000, 600, n_houses).clip(600, 5000)
bedrooms = np.random.choice([1, 2, 3, 4, 5], size=n_houses, p=[0.1, 0.25, 0.4, 0.2, 0.05])
house_age = np.random.randint(0, 80, size=n_houses)
neighborhood_quality = np.random.uniform(1, 10, size=n_houses)

# Calculate house price with some base pricing and feature interactions
base_price = 50000
size_price = size_sqft * 120
bedroom_price = bedrooms * 15000
age_penalty = house_age * 1200
quality_bonus = neighborhood_quality * 25000
noise = np.random.normal(0, 20000, size=n_houses)

house_price = (base_price + size_price + bedroom_price - age_penalty + quality_bonus + noise).clip(30000, 900000)

# Construct the DataFrame
df_houses = pd.DataFrame({
    'Size_SqFt': size_sqft,
    'Bedrooms': bedrooms,
    'House_Age': house_age,
    'Neighborhood_Quality': neighborhood_quality,
    'Price': house_price
})

display(df_houses.head())
```

### 2. Preprocess Data and Train a Gradient Boosting Regressor
We will split the dataset into train and test sets, standardize the feature matrices, and train a powerful Gradient Boosting Regressor.

```python
# Extract features and target variable
X_house = df_houses.drop('Price', axis=1)
y_house = df_houses['Price']

# Split into training and testing sets (80% train, 20% test)
X_train_h, X_test_h, y_train_h, y_test_h = train_test_split(X_house, y_house, test_size=0.2, random_state=42)

# Scale the features
scaler_h = StandardScaler()
X_train_scaled_h = scaler_h.fit_transform(X_train_h)
X_test_scaled_h = scaler_h.transform(X_test_h)

# Train the Gradient Boosting Regressor
gb_house = GradientBoostingRegressor(n_estimators=100, learning_rate=0.1, max_depth=4, random_state=42)
gb_house.fit(X_train_scaled_h, y_train_h)

y_pred_h = gb_house.predict(X_test_scaled_h)
print("Gradient Boosting Model trained successfully!")
```

### 3. Evaluate House Price Predictions
We assess performance using regression metrics (R-squared and MAE) and plot predicted prices against actual market values.

```python
# Compute metrics
mae_h = mean_absolute_error(y_test_h, y_pred_h)
rmse_h = np.sqrt(mean_squared_error(y_test_h, y_pred_h))
r2_h = r2_score(y_test_h, y_pred_h)

print(f"Mean Absolute Error (MAE): ${mae_h:,.2f}")
print(f"Root Mean Squared Error (RMSE): ${rmse_h:,.2f}")
print(f"R-squared Score (R2): {r2_h:.4f}\n")

# Plot predictions vs actuals
plt.figure(figsize=(8, 6))
plt.scatter(y_test_h, y_pred_h, alpha=0.6, color='royalblue')
plt.plot([y_test_h.min(), y_test_h.max()], [y_test_h.min(), y_test_h.max()], 'r--', lw=2)
plt.title('Actual vs Predicted House Prices')
plt.xlabel('Actual Price ($)')
plt.ylabel('Predicted Price ($)')
plt.grid(True)
plt.show()
```
