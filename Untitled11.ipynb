### 1. Create a Synthetic Credit Score Dataset
We will create a mock dataset that mimics realistic features influencing credit scores: Annual Income, Age, Number of Bank Accounts, Number of Credit Cards, Interest Rate, Outstanding Debt, and Credit History Age.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score

# Set seed for reproducibility
np.random.seed(42)
n_samples = 1000

# Generate realistic features
income = np.random.normal(55000, 20000, n_samples).clip(15000, 150000)
age = np.random.randint(18, 70, size=n_samples)
num_cards = np.random.randint(1, 10, size=n_samples)
outstanding_debt = (np.random.exponential(scale=3000, size=n_samples) + (10 - num_cards) * 200).clip(0, 15000)
credit_history_months = (age * 3 + np.random.randint(-12, 12, size=n_samples)).clip(6, 400)

# Define simple logic to assign a Credit Score target ('Poor', 'Standard', 'Good')
score_metric = (income * 0.4) - (outstanding_debt * 1.5) + (credit_history_months * 50) - (num_cards * 300)

credit_score = []
for val in score_metric:
    if val < 15000:
        credit_score.append('Poor')
    elif val < 35000:
        credit_score.append('Standard')
    else:
        credit_score.append('Good')

# Assemble into a pandas DataFrame
df_credit = pd.DataFrame({
    'Age': age,
    'Annual_Income': income,
    'Num_Bank_Accounts': np.random.randint(1, 8, size=n_samples),
    'Num_Credit_Cards': num_cards,
    'Outstanding_Debt': outstanding_debt,
    'Credit_History_Months': credit_history_months,
    'Credit_Score': credit_score
})

display(df_credit.head())
```

### 2. Preprocess Data and Train Classifier
We will encode our target labels, scale our features to have zero mean and unit variance, and train a Random Forest Classifier.

```python
# Split features and target
X_cr = df_credit.drop('Credit_Score', axis=1)
y_cr = df_credit['Credit_Score']

# Encode target variable (Poor: 0, Standard: 1, Good: 2 based on natural progression or alphabet)
label_encoder = LabelEncoder()
y_cr_encoded = label_encoder.fit_transform(y_cr)

# Train-Test Split
X_train_cr, X_test_cr, y_train_cr, y_test_cr = train_test_split(
    X_cr, y_cr_encoded, test_size=0.2, random_state=42, stratify=y_cr_encoded
)

# Feature Scaling
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train_cr)
X_test_scaled = scaler.transform(X_test_cr)

# Train Random Forest Classifier
rf_credit = RandomForestClassifier(n_estimators=100, random_state=42)
rf_credit.fit(X_train_scaled, y_train_cr)

y_pred_cr = rf_credit.predict(X_test_scaled)
print("Model training complete.")
```

### 3. Evaluate Results
We visualize performance using a confusion matrix heatmap and print the detailed classification report.

```python
# Accuracy score
acc_cr = accuracy_score(y_test_cr, y_pred_cr)
print(f"Credit Score Classification Accuracy: {acc_cr:.2%}\n")

# Print classification report
classes = label_encoder.classes_
print("Classification Report:")
print(classification_report(y_test_cr, y_pred_cr, target_names=classes))

# Generate confusion matrix
cm_cr = confusion_matrix(y_test_cr, y_pred_cr)

# Plot matrix
plt.figure(figsize=(7, 5))
sns.heatmap(cm_cr, annot=True, fmt='d', cmap='Purples',
            xticklabels=classes, yticklabels=classes)
plt.title('Credit Score Confusion Matrix')
plt.xlabel('Predicted Class')
plt.ylabel('True Class')
plt.show()
```
