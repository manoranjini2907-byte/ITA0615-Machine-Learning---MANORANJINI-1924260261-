### Candidate-Elimination Algorithm

The **Candidate-Elimination** algorithm computes the version space containing all hypotheses from $H$ that are consistent with a given sequence of training examples.

* **Specific Boundary ($S$):** Initialized to the set of most specific hypotheses in $H$, i.e., $<\emptyset, \emptyset, \dots, \emptyset>$.
* **General Boundary ($G$):** Initialized to the set of most general hypotheses in $H$, i.e., $<?, ?, \dots, ?>$.

```python
import csv
import numpy as np
import pandas as pd

# Create a sample CSV file with training data
csv_data = """Sky,AirTemp,Humidity,Wind,Water,Forecast,EnjoySport
Sunny,Warm,Normal,Strong,Warm,Same,Yes
Sunny,Warm,High,Strong,Warm,Same,Yes
Rainy,Cold,High,Strong,Warm,Change,No
Sunny,Warm,High,Strong,Cool,Same,Yes"""

with open('training_data.csv', 'w') as f:
    f.write(csv_data)

# Load and display the CSV dataset
df = pd.read_csv('training_data.csv')
display(df)
```

```python
def candidate_elimination(concepts, target):
    """
    Implements the Candidate-Elimination algorithm.
    """
    # Initialize Specific and General boundaries
    specific_h = concepts[0].copy()
    generic_h = [["?" for _ in range(len(specific_h))] for _ in range(len(specific_h))]

    for i, h in enumerate(concepts):
        # Positive Example
        if target[i] == "Yes":
            for x in range(len(specific_h)):
                if h[x] != specific_h[x]:
                    specific_h[x] = '?'
                    generic_h[x][x] = '?'

        # Negative Example
        elif target[i] == "No":
            for x in range(len(specific_h)):
                if h[x] != specific_h[x]:
                    generic_h[x][x] = specific_h[x]
                else:
                    generic_h[x][x] = '?'

        # Filter out redundant hypotheses from general boundary
        indices = [idx for idx, val in enumerate(generic_h) if val == ['?'] * len(specific_h)]
        for idx in sorted(indices, reverse=True):
            generic_h.pop(idx)

        print(f"Iteration {i+1}:")
        print(f"Specific Boundary S_{i+1}: {specific_h}")
        print(f"General Boundary G_{i+1}: {generic_h}\n")

    return specific_h, generic_h

# Prepare data matrices
concepts = np.array(df.iloc[:, :-1])
target = np.array(df.iloc[:, -1])

# Run the algorithm
print("--- Running Candidate-Elimination Algorithm ---\n")
s_final, g_final = candidate_elimination(concepts, target)

print("Final Boundaries:")
print(f"S-Boundary: {s_final}")
print(f"G-Boundary: {g_final}")
```
