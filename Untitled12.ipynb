### 1. Load the Iris Dataset and Explore
We will load the classic Iris dataset, which contains 3 classes of iris flowers (Setosa, Versicolor, and Virginica) with 4 features each (sepal length, sepal width, petal length, and petal width).

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score

# Load the Iris dataset
iris = load_iris()
X_iris = iris.data
y_iris = iris.target
feature_names = iris.feature_names
target_names = iris.target_names

# Convert to DataFrame for visualization
df_iris = pd.DataFrame(X_iris, columns=feature_names)
df_iris['species'] = [target_names[i] for i in y_iris]
display(df_iris.head())
```

### 2. Preprocess Data and Train KNN Classifier
We will split the dataset, scale the features using `StandardScaler` to ensure the distance calculations in KNN are unbiased, and then fit the classifier.

```python
# Split into training and testing sets (80% train, 20% test)
X_train_ir, X_test_ir, y_train_ir, y_test_ir = train_test_split(
    X_iris, y_iris, test_size=0.2, random_state=42, stratify=y_iris
)

# Standardize features
scaler_ir = StandardScaler()
X_train_scaled_ir = scaler_ir.fit_transform(X_train_ir)
X_test_scaled_ir = scaler_ir.transform(X_test_ir)

# Initialize and fit KNN Classifier (k=5)
knn_iris = KNeighborsClassifier(n_neighbors=5)
knn_iris.fit(X_train_scaled_ir, y_train_ir)

# Predict on the test set
y_pred_ir = knn_iris.predict(X_test_scaled_ir)
print("KNN model trained successfully on Iris dataset.")
```

### 3. Evaluate the Model
Let's view the accuracy, classification report, and confusion matrix.

```python
# Calculate accuracy
acc_ir = accuracy_score(y_test_ir, y_pred_ir)
print(f"Accuracy: {acc_ir:.2%}\n")

# Display classification report
print("Classification Report:")
print(classification_report(y_test_ir, y_pred_ir, target_names=target_names))

# Plot confusion matrix
cm_ir = confusion_matrix(y_test_ir, y_pred_ir)
plt.figure(figsize=(6, 5))
sns.heatmap(cm_ir, annot=True, fmt='d', cmap='Oranges',
            xticklabels=target_names, yticklabels=target_names)
plt.title('Iris Classification Confusion Matrix')
plt.xlabel('Predicted Species')
plt.ylabel('True Species')
plt.show()
```
