### Naive Bayes Classification for Bank Loan Prediction

This section implements a machine learning workflow using the **Gaussian Naive Bayes** classifier to predict whether a customer will be approved for a bank loan based on demographic and financial characteristics.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import classification_report, confusion_matrix, roc_curve, auc, accuracy_score
```

```python
# 1. Generate synthetic Bank Loan dataset
np.random.seed(42)
n_samples = 1200

data = {
    'Age': np.random.randint(21, 65, size=n_samples),
    'Annual_Income': np.random.normal(65000, 20000, size=n_samples).astype(int),
    'Credit_Score': np.random.randint(300, 851, size=n_samples),
    'Loan_Amount': np.random.normal(150000, 60000, size=n_samples).astype(int),
    'Education': np.random.choice(['Undergrad', 'Graduate', 'Professional'], size=n_samples),
    'Self_Employed': np.random.choice(['Yes', 'No'], size=n_samples, p=[0.15, 0.85])
}

df_loans = pd.DataFrame(data)

# Create a logical loan decision rule with some random noise added
income_ratio = df_loans['Loan_Amount'] / df_loans['Annual_Income']
score_factor = (df_loans['Credit_Score'] - 300) / 550

# Probability of approval base formula
prob_approval = 1 / (1 + np.exp(-(score_factor * 5 - income_ratio * 2 + 1)))
random_noise = np.random.uniform(-0.15, 0.15, size=n_samples)

# Output target: Loan_Approved (1 = Approved, 0 = Denied)
df_loans['Loan_Approved'] = ((prob_approval + random_noise) >= 0.5).astype(int)

display(df_loans.head())
print(f"Target distribution:\n{df_loans['Loan_Approved'].value_counts(normalize=True)}")
```

```python
# 2. Data Preprocessing
# Encode categorical variables
le_edu = LabelEncoder()
df_loans['Education'] = le_edu.fit_transform(df_loans['Education'])

le_emp = LabelEncoder()
df_loans['Self_Employed'] = le_emp.fit_transform(df_loans['Self_Employed'])

# Define features and target
X_loan = df_loans.drop(columns=['Loan_Approved'])
y_loan = df_loans['Loan_Approved']

# Split into training and testing sets (80-20 split)
X_train_l, X_test_l, y_train_l, y_test_l = train_test_split(X_loan, y_loan, test_size=0.20, random_state=42)

# Standardize numerical features for stability
scaler = StandardScaler()
X_train_l_scaled = scaler.fit_transform(X_train_l)
X_test_l_scaled = scaler.transform(X_test_l)
```

```python
# 3. Train Naive Bayes Model
nb_classifier = GaussianNB()
nb_classifier.fit(X_train_l_scaled, y_train_l)

# Predict on the test set
y_pred_l = nb_classifier.predict(X_test_l_scaled)
y_prob_l = nb_classifier.predict_proba(X_test_l_scaled)[:, 1]

# Evaluation Metrics
accuracy_l = accuracy_score(y_test_l, y_pred_l)
print(f"Naive Bayes Test Accuracy: {accuracy_l:.4f}\n")
print("Classification Report:")
print(classification_report(y_test_l, y_pred_l))
```

```python
# 4. Visualization of Performance Metrics
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Confusion Matrix heatmap
cm_loan = confusion_matrix(y_test_l, y_pred_l)
sns.heatmap(cm_loan, annot=True, fmt='d', cmap='Blues', cbar=False,
            xticklabels=['Denied', 'Approved'], yticklabels=['Denied', 'Approved'], ax=axes[0])
axes[0].set_title('Confusion Matrix - Bank Loan Prediction')
axes[0].set_xlabel('Predicted Label')
axes[0].set_ylabel('True Label')

# ROC Curve
fpr, tpr, _ = roc_curve(y_test_l, y_prob_l)
roc_auc = auc(fpr, tpr)

axes[1].plot(fpr, tpr, color='darkorange', lw=2, label=f'ROC Curve (AUC = {roc_auc:.3f})')
axes[1].plot([0, 1], [0, 1], color='navy', lw=2, linestyle='--')
axes[1].set_xlim([0.0, 1.0])
axes[1].set_ylim([0.0, 1.05])
axes[1].set_xlabel('False Positive Rate')
axes[1].set_ylabel('True Positive Rate')
axes[1].set_title('Receiver Operating Characteristic (ROC) Curve')
axes[1].legend(loc="lower right")
axes[1].grid(True, linestyle='--', alpha=0.5)

plt.tight_layout()
plt.show()
```
