### Mobile Price Range Prediction
We will classify mobile phone price ranges (0: Low Cost, 1: Medium Cost, 2: High Cost, 3: Very High Cost) based on phone specifications such as:
* **Battery Power** (mAh)
* **RAM** (MB)
* **Internal Memory** (GB)
* **Processor Cores** (No. of cores)
* **Primary Camera Megapixels** (MP)
* **Screen Dimensions** (Height/Width)
* **4G/5G Connectivity** (Boolean)

We'll use a **Random Forest Classifier** to predict the price ranges, evaluate the classification performance, and plot feature importances.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score

# 1. Generate synthetic Mobile Specifications Dataset
np.random.seed(42)
n_samples = 1500

battery_power = np.random.randint(500, 2000, size=n_samples)
ram = np.random.randint(256, 4096, size=n_samples)
int_memory = np.random.randint(2, 64, size=n_samples)
n_cores = np.random.choice([1, 2, 4, 6, 8], size=n_samples)
pc_megapixels = np.random.randint(2, 20, size=n_samples)
three_g = np.random.choice([0, 1], p=[0.2, 0.8], size=n_samples)
four_g = np.random.choice([0, 1], p=[0.4, 0.6], size=n_samples)

# Calculate a weighted score to assign logical price ranges
price_score = (
    0.4 * (ram / 4096) +
    0.3 * (battery_power / 2000) +
    0.1 * (int_memory / 64) +
    0.1 * (n_cores / 8) +
    0.05 * (pc_megapixels / 20) +
    0.05 * four_g
)

# Bin the score into 4 price ranges (0, 1, 2, 3)
price_range = pd.qcut(price_score, q=4, labels=[0, 1, 2, 3]).astype(int)

df_mobile = pd.DataFrame({
    'Battery_Power': battery_power,
    'RAM': ram,
    'Internal_Memory': int_memory,
    'Cores': n_cores,
    'PC_Megapixels': pc_megapixels,
    '3G': three_g,
    '4G': four_g,
    'Price_Range': price_range
})

display(df_mobile.head())
print(df_mobile['Price_Range'].value_counts())
```

```python
# 2. Prepare Train-Test Split
X_mob = df_mobile.drop(columns=['Price_Range'])
y_mob = df_mobile['Price_Range']

X_train_m, X_test_m, y_train_m, y_test_m = train_test_split(
    X_mob, y_mob, test_size=0.2, random_state=42, stratify=y_mob
)

# Scale numerical variables
scaler_m = StandardScaler()
numerical_cols = ['Battery_Power', 'RAM', 'Internal_Memory', 'PC_Megapixels']

X_train_m_scaled = X_train_m.copy()
X_test_m_scaled = X_test_m.copy()

X_train_m_scaled[numerical_cols] = scaler_m.fit_transform(X_train_m[numerical_cols])
X_test_m_scaled[numerical_cols] = scaler_m.transform(X_test_m[numerical_cols])
```

```python
# 3. Train Random Forest Classifier
mobile_rf = RandomForestClassifier(n_estimators=150, max_depth=10, random_state=42)
mobile_rf.fit(X_train_m_scaled, y_train_m)

y_pred_m = mobile_rf.predict(X_test_m_scaled)

print("Classification Accuracy:", accuracy_score(y_test_m, y_pred_m))
print("\nDetailed Classification Report:\n", classification_report(y_test_m, y_pred_m))
```

```python
# 4. Visualize Confusion Matrix and Feature Importances
fig, axes = plt.subplots(1, 2, figsize=(16, 6))

# Confusion Matrix Plot
cm_m = confusion_matrix(y_test_m, y_pred_m)
sns.heatmap(cm_m, annot=True, fmt='d', cmap='Blues', ax=axes[0],
            xticklabels=['Low', 'Medium', 'High', 'Very High'],
            yticklabels=['Low', 'Medium', 'High', 'Very High'])
axes[0].set_title('Confusion Matrix - Mobile Price Class')
axes[0].set_xlabel('Predicted Label')
axes[0].set_ylabel('True Label')

# Feature Importances Plot
importances = mobile_rf.feature_importances_
indices = np.argsort(importances)[::-1]
feature_names = X_mob.columns

sns.barplot(x=importances[indices], y=[feature_names[i] for i in indices], ax=axes[1], palette='rocket')
axes[1].set_title('Feature Importance for Mobile Price')
axes[1].set_xlabel('Relative Importance')

plt.tight_layout()
plt.show()
```
