### 1. Create a Synthetic Car Price Dataset
We will create a realistic dataset containing continuous and categorical features: **Year of Manufacture**, **Mileage (km)**, **Engine Size (L)**, **Fuel Type** (Petrol/Diesel), and the target variable **Price ($)**.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

# Seed for reproducibility
np.random.seed(42)
n_cars = 1000

# Generate synthetic features
year = np.random.randint(2010, 2023, size=n_cars)
mileage = (2023 - year) * np.random.normal(12000, 3000, size=n_cars).clip(2000, 30000)
engine_size = np.random.choice([1.0, 1.2, 1.4, 1.6, 2.0, 2.5, 3.0], size=n_cars)
fuel_type = np.random.choice(['Petrol', 'Diesel'], size=n_cars, p=[0.7, 0.3])

# Generate a realistic price based on features with some noise
base_price = 25000
depreciation_per_year = (2023 - year) * 1500
mileage_penalty = mileage * 0.08
engine_bonus = (engine_size - 1.0) * 4000
fuel_bonus = np.where(fuel_type == 'Diesel', 1500, 0)
noise = np.random.normal(0, 1500, size=n_cars)

price = (base_price - depreciation_per_year - mileage_penalty + engine_bonus + fuel_bonus + noise).clip(2000, 60000)

# Create DataFrame
df_cars = pd.DataFrame({
    'Year': year,
    'Mileage': mileage,
    'Engine_Size': engine_size,
    'Fuel_Type': fuel_type,
    'Price': price
})

display(df_cars.head())
```

### 2. Preprocess Features and Train Random Forest Regressor
We will build a pipeline that standardizes the numerical features, applies one-hot encoding to our categorical variable (`Fuel_Type`), and trains a `RandomForestRegressor`.

```python
# Define features and target variable
X_cars = df_cars.drop('Price', axis=1)
y_cars = df_cars['Price']

# Split into training and testing sets
X_train_c, X_test_c, y_train_c, y_test_c = train_test_split(X_cars, y_cars, test_size=0.2, random_state=42)

# Define numerical and categorical transformations
num_features = ['Year', 'Mileage', 'Engine_Size']
cat_features = ['Fuel_Type']

preprocessor = ColumnTransformer(
    transformers=[
        ('num', StandardScaler(), num_features),
        ('cat', OneHotEncoder(), cat_features)
    ])

# Build pipeline
car_model_pipeline = Pipeline(steps=[
    ('preprocessor', preprocessor),
    ('regressor', RandomForestRegressor(n_estimators=100, random_state=42))
])

# Train the model
car_model_pipeline.fit(X_train_c, y_train_c)
y_pred_c = car_model_pipeline.predict(X_test_c)

print("Car Price Prediction model trained successfully!")
```

### 3. Evaluate the Prediction Model
We evaluate the model using Mean Absolute Error (MAE), Root Mean Squared Error (RMSE), and the $R^2$ Score, then plot predictions vs actual values.

```python
# Calculate evaluation metrics
mae_c = mean_absolute_error(y_test_c, y_pred_c)
rmse_c = np.sqrt(mean_squared_error(y_test_c, y_pred_c))
r2_c = r2_score(y_test_c, y_pred_c)

print(f"Mean Absolute Error (MAE): ${mae_c:.2f}")
print(f"Root Mean Squared Error (RMSE): ${rmse_c:.2f}")
print(f"R-squared Score (R2): {r2_c:.4f}\n")

# Plot predicted vs actual prices
plt.figure(figsize=(8, 6))
plt.scatter(y_test_c, y_pred_c, alpha=0.6, color='teal')
plt.plot([y_test_c.min(), y_test_c.max()], [y_test_c.min(), y_test_c.max()], 'r--', lw=2)
plt.title('Actual vs Predicted Car Prices')
plt.xlabel('Actual Price ($)')
plt.ylabel('Predicted Price ($)')
plt.grid(True)
plt.show()
```
