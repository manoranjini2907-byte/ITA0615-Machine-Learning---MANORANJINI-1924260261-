### 1. Import Libraries and Generate Continuous Data
Linear Regression is used for predicting continuous values. We will generate a synthetic dataset with a linear relationship and a bit of noise, then split it into training and testing datasets.

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score

# Set random seed for reproducibility
np.random.seed(42)

# Generate synthetic linear data
X_reg = 2 * np.random.rand(100, 1)
y_reg = 4 + 3 * X_reg + np.random.randn(100, 1)

# Split the dataset into training and testing sets
X_train_reg, X_test_reg, y_train_reg, y_test_reg = train_test_split(X_reg, y_reg, test_size=0.2, random_state=42)

# Plot the synthetic data
plt.figure(figsize=(8, 6))
plt.scatter(X_train_reg, y_train_reg, color='blue', label='Train data')
plt.scatter(X_test_reg, y_test_reg, color='orange', label='Test data')
plt.title('Synthetic Dataset for Linear Regression')
plt.xlabel('X (Independent Variable)')
plt.ylabel('y (Dependent Variable)')
plt.legend()
plt.grid(True)
plt.show()
```

### 2. Train the Linear Regression Model
Next, we initialize the `LinearRegression` model from `scikit-learn` and fit it to our training data.

```python
# Initialize the Linear Regression model
lin_reg = LinearRegression()

# Train the model
lin_reg.fit(X_train_reg, y_train_reg)

# Print the learned parameters
print(f"Intercept (beta_0): {lin_reg.intercept_[0]:.4f}")
print(f"Slope/Coefficient (beta_1): {lin_reg.coef_[0][0]:.4f}")
```

### 3. Evaluate the Model
We will evaluate the trained model on our testing dataset using Mean Squared Error (MSE) and the R-squared ($R^2$) score.

```python
# Predict on test data
y_pred_reg = lin_reg.predict(X_test_reg)

# Calculate metrics
mse = mean_squared_error(y_test_reg, y_pred_reg)
r2 = r2_score(y_test_reg, y_pred_reg)

print(f"Mean Squared Error (MSE): {mse:.4f}")
print(f"R-squared (R2 Score): {r2:.4f}")
```

### 4. Plot the Regression Line
To see how well our model fits the data, we will plot the predicted regression line over our original dataset points.

```python
# Generate points for the regression line
X_line = np.linspace(0, 2, 100).reshape(-1, 1)
y_line = lin_reg.predict(X_line)

# Plot regression line over data
plt.figure(figsize=(8, 6))
plt.scatter(X_reg, y_reg, color='blue', alpha=0.6, label='Data points')
plt.plot(X_line, y_line, color='red', linewidth=3, label=f'Best Fit Line: y = {lin_reg.intercept_[0]:.2f} + {lin_reg.coef_[0][0]:.2f}x')
plt.title('Linear Regression Best Fit Line')
plt.xlabel('X')
plt.ylabel('y')
plt.legend()
plt.grid(True)
plt.show()
```
