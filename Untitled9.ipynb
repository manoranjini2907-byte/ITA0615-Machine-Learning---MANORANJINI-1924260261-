### Comparing Linear and Polynomial Regression
We will compare a simple Linear Regression model with a Polynomial Regression model (Degree 3) using our existing regression dataset variables.

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures
from sklearn.metrics import mean_squared_error, r2_score

# --- 1. Linear Regression ---
lin_reg_model = LinearRegression()
lin_reg_model.fit(X_train_reg, y_train_reg)
y_pred_lin = lin_reg_model.predict(X_test_reg)

# --- 2. Polynomial Regression (Degree 3) ---
poly_degree = 3
poly_features = PolynomialFeatures(degree=poly_degree)

# Transform the features
X_train_poly = poly_features.fit_transform(X_train_reg)
X_test_poly = poly_features.transform(X_test_reg)

# Fit the model
poly_reg_model = LinearRegression()
poly_reg_model.fit(X_train_poly, y_train_reg)
y_pred_poly = poly_reg_model.predict(X_test_poly)
```

### Evaluation Metrics Comparison

```python
# Calculate metrics for Linear Regression
mse_lin = mean_squared_error(y_test_reg, y_pred_lin)
r2_lin = r2_score(y_test_reg, y_pred_lin)

# Calculate metrics for Polynomial Regression
mse_poly = mean_squared_error(y_test_reg, y_pred_poly)
r2_poly = r2_score(y_test_reg, y_pred_poly)

print("=== Performance Comparison ===")
print(f"Linear Regression     -> MSE: {mse_lin:.4f}, R2: {r2_lin:.4f}")
print(f"Polynomial (Deg {poly_degree}) -> MSE: {mse_poly:.4f}, R2: {r2_poly:.4f}")
```

### Visualizing the Regression Curves
