### 1. Expectation-Maximization (EM) Algorithm with Gaussian Mixture Models (GMM)

The Expectation-Maximization algorithm is an iterative method used to find maximum likelihood estimates of parameters in probabilistic models with latent variables. The most common application is clustering using Gaussian Mixture Models (GMM).

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs
from sklearn.mixture import GaussianMixture

# Generate synthetic 2D data clusters
X_em, y_true = make_blobs(n_samples=400, centers=4, cluster_std=0.60, random_state=0)

# Plot the data before applying EM
plt.figure(figsize=(8, 6))
plt.scatter(X_em[:, 0], X_em[:, 1], s=40, color='gray', alpha=0.7)
plt.title("Input Data for Expectation-Maximization")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.grid(True)
plt.show()
```

### 2. Fit Gaussian Mixture Model using EM

```python
# Initialize GMM with 4 components
gmm = GaussianMixture(n_components=4, random_state=42)

# Fit the model using the EM algorithm
gmm.fit(X_em)

# Predict the cluster labels
labels = gmm.predict(X_em)

print("EM algorithm converged:", gmm.converged_)
print("Number of iterations to converge:", gmm.n_iter_)
```

### 3. Visualize Clustering and Probability Contours

```python
# Create a grid to plot probabilities contours
x_span = np.linspace(X_em[:, 0].min() - 1, X_em[:, 0].max() + 1, 100)
y_span = np.linspace(X_em[:, 1].min() - 1, X_em[:, 1].max() + 1, 100)
XX, YY = np.meshgrid(x_span, y_span)
XX_YY_flat = np.c_[XX.ravel(), YY.ravel()]

# Calculate log likelihood of each point under the GMM model
Z_scores = -gmm.score_samples(XX_YY_flat)
Z_scores = Z_scores.reshape(XX.shape)

# Plot
plt.figure(figsize=(10, 8))
plt.contour(XX, YY, Z_scores, norm=None, levels=np.logspace(0, 2, 10), colors='black', alpha=0.3)
plt.scatter(X_em[:, 0], X_em[:, 1], c=labels, s=40, cmap='viridis', zorder=2, edgecolors='k')
plt.title("GMM Clustering via Expectation-Maximization")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.colorbar(label='Cluster Assignment')
plt.grid(True)
plt.show()
```
